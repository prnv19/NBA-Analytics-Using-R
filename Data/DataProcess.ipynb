```python
import pandas as pd
from unidecode import unidecode

# Load your dataset
df_norm = pd.read_csv('Raw/nba_normal_scorers.csv')
df_out = pd.read_csv('Raw/nba_high_scorers_outliers.csv')

# Step 1: Identify players who have a '2TM' row
players_with_2tm_norm = df_norm[df_norm['Team'] == '2TM']['Player'].unique()
players_with_2tm_out = df_out[df_out['Team'] == '2TM']['Player'].unique()

# Step 2: Keep only:
# - rows where Team == '2TM'
# - OR rows where the player does not appear in the 2TM list
df_filtered_norm = df_norm[(df_norm['Team'] == '2TM') | (~df_norm['Player'].isin(players_with_2tm_norm))]
df_filtered_out = df_out[(df_out['Team'] == '2TM') | (~df_out['Player'].isin(players_with_2tm_out))]

# Reset index
df_filtered_norm = df_filtered_norm.reset_index(drop=True)
df_filtered_out = df_filtered_out.reset_index(drop=True)
```

Combine salary

```python
df_salary = pd.read_csv('Raw/nba_player_salaries_by_year.csv')
df_salary
```

```python
for df in (df_filtered_norm, df_filtered_out, df_salary):
    df['Player'] = df['Player'].astype(str).apply(lambda x: unidecode(x).strip())
    df['Year']   = df['Year'].astype(int)

# 3. Merge Salary (keep only the numeric Salary column from df_sal)
salary_cols = ['Player', 'Year', 'Salary']
df_norm_merged = pd.merge(df_filtered_norm, df_salary[salary_cols], on=['Player', 'Year'], how='left')
df_out_merged  = pd.merge(df_filtered_out,  df_salary[salary_cols], on=['Player', 'Year'], how='left')
```

```python
# Save the filtered dataset

# Separate data for the year 2025
df_norm_2025 = df_norm_merged[df_norm_merged['Year'] == 2025].drop(columns=['Salary'])
df_out_2025 = df_out_merged[df_out_merged['Year'] == 2025].drop(columns=['Salary'])

# Save the 2025 data separately
df_norm_2025.to_csv('Processed/nba_normal_scorers_2025.csv', index=False)
df_out_2025.to_csv('Processed/nba_high_scorers_outliers_2025.csv', index=False)

# Remove salary for the year 2025 from the original datasets
df_norm_merged = df_norm_merged[df_norm_merged['Year'] != 2025]
df_out_merged = df_out_merged[df_out_merged['Year'] != 2025]

df_norm_merged.to_csv('Processed/nba_normal_scorers_filtered.csv', index=False)
df_out_merged.to_csv('Processed/nba_high_scorers_outliers_filtered.csv', index=False)
```
